```python
import sqlite3

import pandas as pd
from bertopic import BERTopic

conn = sqlite3.connect("db/axiom.db")  # Creates or opens the database file
cursor = conn.cursor()

cursor.execute('''
    SELECT speech, summary
    FROM speeches s
    ORDER BY date
''')

data = cursor.fetchall()
conn.close()

print(len(data))
```

```python
speeches = []
summaries = []
for entry in data:
    speeches.append(entry[0])
    summaries.append(entry[1])

topic_model = BERTopic()
topics, probs = topic_model.fit_transform(summaries)

df = topic_model.get_topic_info()
print(df)

data = cursor.fetchmany(100)
```
