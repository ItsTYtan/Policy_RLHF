```python
import sqlite3

import pandas as pd
from bertopic import BERTopic

conn = sqlite3.connect("db/axiom.db")  # Creates or opens the database file
cursor = conn.cursor()

cursor.execute('''
    SELECT speech, summary
    FROM speeches s
    ORDER BY date
''')

data = cursor.fetchall()
conn.close()

print(len(data))
```

```python
speeches = []
claims = []
for entry in data:
    speeches.append(entry[0])
    claims.append(entry[1])

base = BERTopic()
base_topics, base_probs = base.fit_transform(claims[0:7000])

new = BERTopic()
new_topics, new_probs = new.fit_transform(claims[7000:8975])

merged_model = BERTopic.merge_models([base, new])
```

```python
print(base.get_topic_info())
```

```python
topic0claims = [claim for topic, claim in list(zip(base_topics, claims[0:7000])) if topic == 0]
print(topic0claims)
```

```python
recursiveModel = BERTopic()
recursive_topics, recursive_probs = recursiveModel.fit_transform(topic0claims)
```
